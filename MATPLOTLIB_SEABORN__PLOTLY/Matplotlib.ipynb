```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title('Hello')
plt.xlabel('X')
plt.ylabel('Y')
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,1)
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
```

```python
plt.subplot(2,2,1)
plt.plot(x**2,x**2)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x**2,x**2)
```

```python
plt.subplot(6,6,1)
plt.subplot(6,6,2)
plt.subplot(6,6,3)
plt.subplot(6,6,4)
plt.subplot(6,6,5)
plt.subplot(6,6,6)
plt.subplot(6,6,7)
```

**Object Oriented way**

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3, 0.3, 0.5, 0.5])
axis2.plot(x,y)
```

```python
fig = plt.figure(figsize = (12,8), dpi = 100)
fig = plt.figure()
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3, 0.3, 0.5, 0.5])
axis2.plot(x,y)
```
